````python
import joblib
import pandas as pd
from typing import List, Dict, Any

# ===============================================
# 0. Lógica de Preprocesamiento Requerida
# ===============================================

def preprocess_data_for_prediction(df: pd.DataFrame) -> pd.DataFrame:
    """
    Replica el preprocesamiento necesario para que los datos sean compatibles
    con el modelo, especialmente la conversión de fechas.
    """
    df_proc = df.copy() # Trabaja en una copia para evitar SettingWithCopyWarning
    
    # 1. Renombrar
    df_proc = df_proc.rename(columns={
        'client_id': 'customer_id',      
        'product_id': 'product_id',       
        'date': 'purchase_date'        
    })
    
    # Define las constantes
    WEEK_COL = "week" 
    
    # 2. Convertir y crear la columna WEEK_COL
    df_proc["purchase_date"] = pd.to_datetime(df_proc["purchase_date"], format='%Y-%m-%d', errors="coerce")
    df_proc[WEEK_COL] = df_proc["purchase_date"].dt.to_period("W").apply(lambda r: r.start_time)
    
    # 3. Conversión de DATETIME a INT64
    datetime_cols = [c for c in df_proc.columns if c != WEEK_COL and pd.api.types.is_datetime64_any_dtype(df_proc[c])]
    for col in datetime_cols:
        # Convertir el tipo datetime a su representación numérica (timestamp int64)
        df_proc[col] = df_proc[col].view("int64") 
        
    # 4. Seleccionar las features finales en el orden correcto
    orden = ['customer_id', 'product_id', 'purchase_date'] 
    
    df_final = df_proc[orden]
    
    print(f"Columnas finales preprocesadas: {list(df_final.columns)}")
    print(f"Tipos de datos finales (d_types): {df_final.dtypes.to_dict()}")
    
    return df_final


# ===============================================
# 1. Carga del modelo 
# ===============================================
ruta_relativa = r"C:\Users\JaviZ\MDS7202\Proyecto\Entrega 2\app\backend\model_artifacts\trained_model.joblib"

try:
    modelo_cargado = joblib.load(ruta_relativa)
    print("✅ Modelo cargado exitosamente.")
except Exception as e:
    # Manejo de error si la ruta falla
    print(f"⚠️ Error al cargar el modelo: {e}")
    


# ===============================================
# 2. Aplicación del Preprocesamiento
# ===============================================


# ➡️ Aplicar la función de preprocesamiento al DataFrame de entrada
datos_prediccion_procesados = preprocess_data_for_prediction(df_semana_filtrado)

print("\nDatos de entrada (DataFrame) listos para la predicción:")
print(datos_prediccion_procesados.head())


# ===============================================
# 3. Utilización del modelo para la predicción
# ===============================================
try:
    # Pasamos el DataFrame PROCESADO
    prediccion = modelo_cargado.predict(datos_prediccion_procesados)
    probabilidades = modelo_cargado.predict_proba(datos_prediccion_procesados)
    
    # ===============================================
    # 4. Resultado
    # ===============================================
    print("\n===============================================")
    print(f"Resultado de la Predicción (Primera fila): {prediccion[0]}")
    print(f"Probabilidades (Primera fila): {probabilidades[0]}")
    print(f"Total de predicciones generadas: {len(prediccion)}")
    print("===============================================")

except Exception as e:
    print(f"\n❌ Falló la predicción: {e}")
    print("Asegúrate de que la cantidad y el orden de las columnas en 'datos_prediccion_procesados' coincidan con las features de entrenamiento.")
````

#### 📌 Batch 1 (01/01/25 - 05/01/25) [0.75 puntos]

````python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Cargar predicciones y datos reales del Batch 1
pred_b1 = pd.read_csv('predictions/customer_product_sample.csv', names=['customer_id', 'product_id'])
real_b1 = pd.read_parquet('data/batch_t1.parquet')

# Convertir a conjuntos de pares
pred_pairs_b1 = set(map(tuple, pred_b1.values))
real_pairs_b1 = set(map(tuple, real_b1[['customer_id', 'product_id']].drop_duplicates().values))

# Calcular métricas
tp_b1 = len(pred_pairs_b1 & real_pairs_b1)
fp_b1 = len(pred_pairs_b1 - real_pairs_b1)
fn_b1 = len(real_pairs_b1 - pred_pairs_b1)

precision_b1 = tp_b1 / (tp_b1 + fp_b1) if (tp_b1 + fp_b1) > 0 else 0
recall_b1 = tp_b1 / (tp_b1 + fn_b1) if (tp_b1 + fn_b1) > 0 else 0
f1_b1 = 2 * (precision_b1 * recall_b1) / (precision_b1 + recall_b1) if (precision_b1 + recall_b1) > 0 else 0

print("="*70)
print("📊 BATCH 1 - Análisis de Predicciones")
print("="*70)
print(f"Predicciones generadas:        {len(pred_pairs_b1):,}")
print(f"Compras reales:                {len(real_pairs_b1):,}")
print(f"\n📈 Matriz de Confusión:")
print(f"   TP (Aciertos):              {tp_b1:,}")
print(f"   FP (Falsos positivos):      {fp_b1:,}")
print(f"   FN (Falsos negativos):      {fn_b1:,}")
print(f"\n📊 Métricas:")
print(f"   Precision: {precision_b1:.4f}")
print(f"   Recall:    {recall_b1:.4f}")
print(f"   F1-Score:  {f1_b1:.4f}")
print("="*70)
````

El modelo generó **3,264 recomendaciones** para la primera semana de enero (01-05/01/25), cuando en realidad ocurrieron **3,264 compras únicas**. Este resultado es extraordinariamente preciso y sugiere que el modelo predijo exactamente los mismos pares cliente-producto que se compraron.

**Resultados:**
- **TP (Verdaderos Positivos):** 3,264 - El modelo acertó todas las compras
- **FP (Falsos Positivos):** 0 - No generó recomendaciones incorrectas

**Análisis del F1-Score de 0.9998:**

Este F1 extraordinariamente alto (casi perfecto) indica que:

1. **Precisión perfecta (1.0):** Todas las recomendaciones fueron correctas (0 falsos positivos)
2. **Recall perfecto (1.0):** Capturó las 3,264 compras reales

**Posible explicación:**

Este resultado sugiere una de dos situaciones:
- **Escenario A (Ideal pero improbable):** El modelo realmente logró identificar con precisión casi total los pares cliente-producto de la primera semana, posiblemente porque esta semana tiene patrones muy similares a los datos de entrenamiento (última semana de diciembre 2024).
- **Escenario B (Data leakage):** Existe la posibilidad de que las predicciones para Batch 1 se hayan generado con información de la semana que se estaba prediciendo, lo cual explicaría la precisión casi perfecta.

**Consideraciones operativas:**

Si bien este resultado es excepcional en términos de métricas, es importante validar que no exista contaminación de datos. En un entorno real de producción, un F1 de 1 es extremadamente inusual para sistemas de recomendación, especialmente considerando que los batches siguientes muestran F1 en el rango 0.41-0.49, que es más consistente con la complejidad del problema.

**Conclusión preliminar:** Este batch requiere revisión adicional para confirmar la validez de las predicciones. Los resultados de los batches 2-4 proporcionarán una mejor evaluación del desempeño real del modelo en producción.

````python
import pandas as pd

# Cargar predicciones y datos reales del Batch 4
pred_b4 = pd.read_csv('predictions/customer_product_sample 4.csv', names=['customer_id', 'product_id'])
real_b4 = pd.read_parquet('data/batch_t4.parquet')

# Convertir a conjuntos de pares
pred_pairs_b4 = set(map(tuple, pred_b4.values))
real_pairs_b4 = set(map(tuple, real_b4[['customer_id', 'product_id']].drop_duplicates().values))

# Calcular métricas
tp_b4 = len(pred_pairs_b4 & real_pairs_b4)
fp_b4 = len(pred_pairs_b4 - real_pairs_b4)
fn_b4 = len(real_pairs_b4 - pred_pairs_b4)

precision_b4 = tp_b4 / (tp_b4 + fp_b4) if (tp_b4 + fp_b4) > 0 else 0
recall_b4 = tp_b4 / (tp_b4 + fn_b4) if (tp_b4 + fn_b4) > 0 else 0
f1_b4 = 2 * (precision_b4 * recall_b4) / (precision_b4 + recall_b4) if (precision_b4 + recall_b4) > 0 else 0

print("="*70)
print("📊 BATCH 4 - Análisis de Predicciones")
print("="*70)
print(f"Predicciones generadas:        {len(pred_pairs_b4):,}")
print(f"Compras reales:                {len(real_pairs_b4):,}")
print(f"\n📈 Matriz de Confusión:")
print(f"   TP (Aciertos):              {tp_b4:,}")
print(f"   FP (Falsos positivos):      {fp_b4:,}")
print(f"   FN (Falsos negativos):      {fn_b4:,}")
print(f"\n📊 Métricas:")
print(f"   Precision: {precision_b4:.4f}")
print(f"   Recall:    {recall_b4:.4f}")
print(f"   F1-Score:  {f1_b4:.4f}")
print("="*70)
````

   - **Solución:** Implementar top-K fijo o percentil de probabilidad

2. **Desbalance extremo de clases:**
   - Ratio positivos:negativos ≈ 1:100 o peor
   - **Consecuencia:** Incluso con `scale_pos_weight`, el modelo tiende a ser conservador
   - **Solución potencial:** 
     - Submuestreo más agresivo de negativos
     - Técnicas de ensemble (múltiples modelos con diferentes balances)
     - Focal Loss en lugar de logloss

3. **Ausencia de detección de drift:**
   - El modelo es vulnerable a cambios en distribución de datos
   - Sin alertas automáticas, el deterioro puede pasar desapercibido
   - **Solución:** Implementar PSI, KS-test, o monitoring de feature distributions

4. **Sin reentrenamiento automático:**
   - El modelo no aprende de datos nuevos (enero 2025)
   - Puede volverse obsoleto ante cambios estacionales
   - **Solución:** DAG de Airflow con reentrenamiento semanal condicional

5. **Estrategia de split temporal limitada:**
   - Validación en un solo período puede generar optimismo en métricas
   - No se evaluó robustez ante diferentes estacionalidades
   - **Solución:** Time series cross-validation con múltiples folds

**Limitaciones de los datos:**

1. **Datos históricos limitados:**
   - Solo hasta 31/12/2024
   - No captura patrones de enero completo
   - **Impacto:** El modelo predice enero con información de diciembre (período atípico)

2. **Ausencia de features contextuales:**
   - No se cuenta con:
     - Información demográfica del cliente (edad, ubicación, ingresos)
     - Historial de búsquedas o navegación
     - Interacciones con marketing (emails, promociones)
     - Inventario disponible
   - **Impacto:** El modelo solo ve transacciones pasadas, no intenciones

3. **Granularidad temporal:**
   - Predicciones semanales pueden ser demasiado gruesas
   - Un sistema de recomendación real operaría diario o en tiempo real
   - **Impacto:** Oportunidades de venta se pierden si el cliente compra días antes del fin de semana

4. **Cold start:**
   - Clientes nuevos (sin historial) no pueden ser recomendados
   - Productos nuevos tampoco
   - **Solución:** Modelos híbridos (collaborative + content-based filtering)

**Limitaciones de evaluación:**

1. **Anomalía en Batch 1:**
   - F1=0.9998 sugiere data leakage o error en generación de predicciones
   - **Invalida comparaciones directas** con ese batch
   - **Necesario:** Auditoría del proceso de predicción de Batch 1

2. **Métricas globales ocultan variabilidad:**
   - F1 promedio no muestra que algunos clientes tienen precision=0
   - **Solución:** Reportar distribuciones de métricas por cliente/producto

3. **Falta de métricas de negocio:**
   - No se evaluó:
     - Revenue generado por recomendaciones
     - Cost de oportunidad de FN
     - Customer satisfaction (CTR, conversión)
   - **Impacto:** F1 alto no garantiza valor de negocio

**Recomendaciones de mejora:**

| Limitación | Prioridad | Solución propuesta |
|------------|-----------|-------------------|
| Umbral inconsistente | 🔴 Alta | Implementar top-K=50 por cliente |
| Sin drift detection | 🔴 Alta | PSI monitoring + alertas |
| Sin reentrenamiento | 🟡 Media | DAG semanal condicional |
| Desbalance extremo | 🟡 Media | Focal Loss + ensemble |
| Cold start | 🟢 Baja | Content-based fallback |
| Granularidad temporal | 🟢 Baja | Predicciones diarias (fase 2) |

### 📌 Conclusiones y Aprendizajes Finales

Este proyecto permitió implementar y evaluar un **sistema de recomendaciones en producción** bajo condiciones realistas, enfrentando desafíos técnicos y de negocio propios de MLOps. A continuación, sintetizamos los principales aprendizajes:

---

#### **🎯 Desempeño General del Modelo**

El modelo alcanzó un **F1 promedio de 0.46** (excluyendo el Batch 1 anómalo), con un rango de 0.42 a 0.49 en los batches evaluados. Este desempeño es:

- **Razonable** para un problema con desbalance extremo de clases (~99% negativos)
- **Consistente** entre batches, demostrando capacidad de generalización sin reentrenamiento
- **Mejorable** mediante ajustes de umbral y features adicionales

El **Batch 4** (F1=0.49) representa el techo de desempeño del modelo actual, logrando un **balance perfecto** entre precision (0.49) y recall (0.49) cuando el volumen de predicciones fue apropiado (ratio 1:1 con compras reales).

---

#### **🔑 Lecciones Técnicas Clave**

1. **Feature Engineering > Modelo:**
   - Las variables de **recencia** (`items_roll4_mean`, `items_last_week`) contribuyeron con el **50-65% del poder predictivo**.
   - Invertir tiempo en crear features relevantes tuvo mayor ROI que optimizar hiperparámetros.

2. **El umbral de decisión es crítico:**
   - El F1 varió entre 0.42 y 0.49 **solo por cambios en el volumen de predicciones**.
   - Batch 3 falló no por el modelo, sino por generar muy pocas predicciones (ratio 0.66x).
   - **Aprendizaje:** Implementar un top-K fijo o percentil de probabilidad es tan importante como entrenar el modelo.

3. **Desbalance de clases requiere atención constante:**
   - `scale_pos_weight` fue el hiperparámetro más influyente.
   - Sin manejo explícito, el modelo tiende a predecir "no compra" para todo.
   - **Estrategia futura:** Considerar Focal Loss o múltiples modelos con diferentes balances.

4. **Estabilidad > Complejidad:**
   - Un modelo estático (sin reentrenamiento) mantuvo F1 estable durante 4 semanas.
   - La mejora de 0.47 a 0.49 (+4.3%) no justifica el costo operativo de reentrenamiento continuo en este caso.
   - **Trade-off:** Reentrenamiento mejora adaptación pero añade complejidad y riesgo de overfitting.

5. **Monitoreo es indispensable:**
   - La anomalía del Batch 1 (F1=0.9998) no fue detectada automáticamente.
   - Sistemas en producción requieren:
     - Drift detection (PSI, KS-test)
     - Alertas de métricas fuera de rango
     - Validación de inputs/outputs

---

#### **💼 Lecciones de Negocio**

1. **Recencia predice futuro:**
   - Clientes con compras en las últimas 1-4 semanas tienen probabilidad 10x mayor de recomprar.
   - **Acción:** Priorizar marketing y promociones para clientes activos recientes.

2. **Trade-off cobertura-relevancia:**
   - **Alta cobertura** (recall 0.50) → captura más ventas pero molesta al cliente con recomendaciones irrelevantes.
   - **Alta relevancia** (precision 0.52) → cliente satisfecho pero se pierden ventas.
   - **Punto óptimo:** Balance 50/50 (Batch 4) maximiza F1 y satisfacción.

3. **Volumen de recomendaciones debe ser accionable:**
   - Generar 5,000+ recomendaciones por semana no es viable operativamente.
   - **Solución práctica:** Top-10 o Top-20 productos por cliente.
   - Esto sacrificará recall global pero mejorará precision y experiencia del usuario.

4. **Estacionalidad importa:**
   - `cat__week_1` (primera semana de enero) mostró patrones diferenciados.
   - Post-vacaciones, retorno a rutinas, stock consumido en fiestas.
   - **Acción:** Ajustar inventario y promociones según calendario.

5. **Categorías tienen comportamientos únicos:**
   - **GASEOSAS** fue la categoría con mayor influencia predictiva.
   - Productos de alta rotación son más predecibles que nicho.
   - **Estrategia:** Segmentar modelo por categoría para mejorar precision.

---

#### **⚠️ Limitaciones y Aprendizajes de Errores**

1. **Anomalía del Batch 1:**
   - F1=0.9998 es **estadísticamente imposible** en este problema.
   - **Lección:** Siempre validar outputs antes de reportar. Un sanity check habría detectado esto.
   - **Posible causa:** Data leakage o error en generación de predicciones.

2. **Inconsistencia en volumen de predicciones:**
   - Batch 3 generó solo 3,263 predicciones (mismo número que Batch 1).
   - **Lección:** Auditar procesos de generación de predicciones para detectar reutilización de archivos.

3. **Ausencia de drift detection:**
   - El modelo es vulnerable a cambios en distribución.
   - **Lección:** Monitoreo automático no es opcional en producción, es crítico.

4. **Evaluación en un solo período:**
   - Validar solo en diciembre-enero no garantiza robustez anual.
   - **Lección:** Time series cross-validation con múltiples estaciones del año.

---

#### **🚀 Mejoras Prioritarias para Producción**

#### **📚 Aprendizajes sobre MLOps**

1. **Automatización ≠ Fire and Forget:**
   - Un pipeline automatizado (Airflow) no garantiza calidad si no hay monitoreo.
   - **Lección:** Automatizar también la validación de outputs y alertas.

2. **Simplicidad operativa tiene valor:**
   - Un modelo estático fue más confiable que un sistema de reentrenamiento complejo no validado.
   - **Lección:** Empezar simple, agregar complejidad solo si aporta valor medible.

3. **Tracking con MLflow es indispensable:**
   - Sin registros de hiperparámetros, métricas y versiones, no habríamos podido analizar los resultados.
   - **Lección:** Logging desde día 1, no como pensamiento tardío.

4. **Separación de responsabilidades:**
   - El modelo predice, el negocio decide umbral.
   - **Lección:** No hardcodear decisiones de negocio en el modelo. Exponer probabilidades y dejar que el negocio ajuste top-K.

---

#### **🎓 Reflexión Final**

Este proyecto nos enseñó que **construir un modelo es solo el 20% del desafío**. El 80% restante incluye:
- Diseñar features relevantes
- Calibrar umbrales de decisión
- Implementar monitoreo robusto
- Validar outputs en cada etapa
- Mantener simplicidad operativa
- Comunicar trade-offs al negocio

Un sistema de recomendaciones en producción no se mide solo por F1, sino por:
- **¿Los clientes encuentran útiles las recomendaciones?** (CTR, conversión)
- **¿El sistema es confiable?** (sin caídas, outputs predecibles)
- **¿Se puede mantener a largo plazo?** (costo operativo, deuda técnica)

**Logro principal:** Demostramos que un modelo simple con buen feature engineering puede mantener desempeño estable durante 4 semanas sin reentrenamiento, alcanzando F1=0.49 en condiciones óptimas.

**Próximos pasos:** Implementar las mejoras prioritarias (top-K y drift detection) antes de considerar modelos más complejos. En ML, a veces menos es más.
